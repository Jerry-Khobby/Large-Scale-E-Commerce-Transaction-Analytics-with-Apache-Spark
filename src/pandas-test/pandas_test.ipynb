```python
import pandas as pd 
import logging 
import os
import time 
import sys 
```

```python
LOG_DIR = "../../logs"
LOG_FILE = os.path.join(LOG_DIR,'pandas_test.log')
os.makedirs(LOG_DIR,exist_ok=True)
```

```python
logger = logging.getLogger("PandasTest")
logger.setLevel(logging.INFO)

formatter = logging.Formatter("%(asctime)s | %(levelname)s | %(message)s")

file_handler = logging.FileHandler(LOG_FILE)
file_handler.setFormatter(formatter)

console_handler = logging.StreamHandler()
console_handler.setFormatter(formatter)

logger.addHandler(file_handler)
logger.addHandler(console_handler)
```

```python
def load_csv_with_timing(file_path, **read_csv_kwargs):
    """
    Loads a CSV file into a Pandas DataFrame, measuring load time and handling MemoryError.
    
    Parameters:
        file_path (str): Path to the CSV file.
        **read_csv_kwargs: Additional keyword arguments for pandas.read_csv().
    
    Returns:
        tuple: (DataFrame or None, load_time_in_seconds or None)
    """
    if not os.path.isfile(file_path):
        logger.info(f"File path {file_path} does not exist")
        return None, None

    start_time = time.perf_counter()

    try:
        df = pd.read_csv(file_path, **read_csv_kwargs)
        load_time = time.perf_counter() - start_time
        logger.info(f"Data loaded successfully in {load_time:.2f} seconds")
        logger.info(f"DataFrame size: {sys.getsizeof(df)/(1024**2):.2f} MB")
        return df, load_time

    except MemoryError:
        load_time = time.perf_counter() - start_time
        logger.info("MemoryError: Could not load file due to insufficient RAM")
        logger.info(f"Time before failure: {load_time:.2f} seconds")
        return None, load_time

    except pd.errors.EmptyDataError:
        logger.info("Error: The file is empty.")
        return None, None

    except pd.errors.ParserError as e:
        logger.info(f"Parsing Error: {e}")
        return None, None

    except Exception as e:
        logger.info(f"Unexpected error: {e}")
        return None, None
```

```python
file_path = "../../data/dataset.csv"
df,load_time = load_csv_with_timing(file_path)

```
